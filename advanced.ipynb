```python
def match_point(point:tuple):
    match point:
        case (0, 0):
            print("Origin")
        case (x, 0):
            print(f"on x-axis, {x=}")
        case (0, y):
            print(f"on y-axis, {y=}")
        case (x, y):
            print(f"{x=}, {y=}")
        case others:
            print(f"{others} is not a valid point")
p1 = (0, 2)
p2 = (2, 0)
match_point(p1)
```
